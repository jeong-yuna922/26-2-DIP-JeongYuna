```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# 사진 불러오기
img = Image.open('/content/mu.jpg')

# numpy 배열로 변환
img_array = np.array(img)

# 크기와 데이터 타입 확인
print("Image shape:", img_array.shape)
print("Data type:", img_array.dtype)

# 이미지 출력
plt.imshow(img_array)
plt.axis('off')
plt.show()
```
